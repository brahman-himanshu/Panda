```python
pivot_data = {
    'Date': [
        '2023-01-01','2023-01-02','2023-01-03','2023-01-04','2023-01-05',
        '2023-01-06','2023-01-07','2023-01-08','2023-01-09','2023-01-10',
        '2023-01-11','2023-01-12','2023-01-13','2023-01-14','2023-01-15',
        '2023-01-16','2023-01-17','2023-01-18','2023-01-19','2023-01-20'
    ],

    'Product': [
        'A','B','C','D','A',
        'B','C','D','A','B',
        'C','D','A','B','C',
        'D','A','B','C','D'
    ],

    'Region': [
        'East','West','North','South','East',
        'West','North','South','East','West',
        'North','South','East','West','North',
        'South','East','West','North','South'
    ],

    'Sales': [
        450, 780, 320, 910, 560,
        670, 230, 850, 490, 720,
        380, 640, 590, 810, 270,
        930, 410, 760, 350, 880
    ],

    'Units': [
        25, 45, 18, 52, 31,
        39, 15, 48, 28, 42,
        21, 35, 33, 46, 17,
        55, 24, 41, 20, 50
    ],

    'Rep': [
        'John','Mary','Bob','Alice','John',
        'Mary','Bob','Alice','John','Mary',
        'Bob','Alice','John','Mary','Bob',
        'Alice','John','Mary','Bob','Alice'
    ]
}

pdf = pd.DataFrame(pivot_data)
```

```python
def extracted_episodes(txt):
    start = txt.find('(')
    end = txt.find(')', start)

    if start != -1 and end != -1:
        episode = txt[start + 1:end].split()[0]
        if episode.isdigit():
            return int(episode)

    return None
```

```python
dff["Episode"] = dff["Title"].apply(extracted_episodes)
```

```python
dff[["Title", "Episode"]].head(10)
```

```python
dff["Episode"] = dff["Episode"].astype(int)
```

```python
dff.columns 
```

```python
dff
```

```python
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1, i+20):
                data += txt[j]

            return data
```

```python
dff['Total Time'] = dff['Title'].apply(extraction_time)

dff.head()
```
